# **Mínimos Quadrados - Água dos Alpes**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
# 1. Carregar Dataset
df = pd.read_csv('agua_alpes.csv')

bpt = df[['BPt']]
pressure = df['Pressure']

#2. Regressão
regressao = linear_model.LinearRegression()
regressao.fit(bpt, pressure)

coef = regressao.coef_
intercept = regressao.intercept_

print("Coef Angular:", coef, "------- Coef Linear:", intercept)
previsao = regressao.predict(bpt)
plt.scatter(bpt, pressure, color="red")
plt.plot(bpt, previsao)

#3. Prevendo Valores
entrada = pd.DataFrame([[200.0], [215.0]], columns=['BPt'])
print("Previsão para valores 200.0 e 215.0", regressao.predict(entrada), "\n\n")
```

# **Mínimos Quadrados - Livros e Aulas x Notas**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
# 1. Carregar Dataset
df = pd.read_csv('notas.csv')

livros_aulas = df[['livros', 'aulas']]
notas = df['notas']

#2. Regressão
regressao = linear_model.LinearRegression()
regressao.fit(livros_aulas, notas)

coef = regressao.coef_
intercept = regressao.intercept_

print("Coef Angular:", coef, "------- Coef Linear:", intercept)
previsao = regressao.predict(livros_aulas)

#3. Prevendo Valores
print("Previsão para 2 livros e 11 aulas:", regressao.predict([[2, 11]]))
print("Previsão para 0 livros e 5 aulas:", regressao.predict([[0, 5]]))
print("Previsão para 4 livros e 20 aulas:", regressao.predict([[4, 20]]))
print("Previsão para 2 livros e 10 aulas:", regressao.predict([[2, 10]]))
print("Previsão para 4 livros e 15 aulas:", regressao.predict([[4, 20]]), "\n\n")

```

# **KNN - Classificação *Iris Flower***

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Carregar o dataset Iris
iris = load_iris()
X = iris.data  # Características: Comprimento e largura da sépala e pétala
y = iris.target  # Classes: Virginica, Setosa e Versicolor

# 2. Dividir os dados em conjuntos de treinamento e teste (ex: 70% treino, 30% teste)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 3. Criar e treinar o modelo KNN (KNeighborsClassifier)
# k=3 é um valor comum para inicialização
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

# 4. Fazer previsões no conjunto de teste
y_pred = knn.predict(X_test)

# 5. Avaliar o desempenho do modelo
accuracy = accuracy_score(y_test, y_pred)
print(f"Acurácia do modelo: {accuracy * 100:.2f}%\n")

# Exibir relatório detalhado com nomes das espécies
print("Relatório de Classificação:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))

```
